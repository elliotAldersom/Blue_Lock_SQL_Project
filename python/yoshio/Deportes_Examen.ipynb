### `df.groupby(args)`
- Método de particionamiento y agregación, permite segmentar datos según claves categóricas y computar resúmenes por grupo.
- Permite calcular métricas por segmento (p. ej. goles totales por equipo, promedio de minutos por posición) de forma vectorizada. 

```python
df_sorted_equipos_asc.head()
```

```python
df_groupby_equipo = df.groupby(by='equipo', as_index=False) # as_index = False para que el label 'equipo' sea columna
type(df_groupby_equipo)
```

```python
# Ver cantidad de rows que tiene el DataframeGroupBy
df_groupby_equipo['goles'].mean()
```

```python
# usar .agg(args) con groupby 
df_estadisticas_equipo = df.groupby(by='equipo').agg(
    # totales
    total_apariciones=('equipo', 'size'),
    total_minutos=('minutos_jugados', 'sum'), 
    total_horas=('horas_jugadas', 'sum'),
    total_goles=('goles', 'sum'), 
    total_asistencias=('asistencias', 'sum'), 
    total_tarjestas=('tarjetas' , 'sum'), 

    # promedios
    promedio_minutos=('minutos_jugados', 'mean'), 
    promedio_horas=('horas_jugadas', 'mean'),
    promedio_goles=('goles', 'mean'), 
    promedio_asistencias=('asistencias', 'mean'), 
    promedio_tarjestas=('tarjetas' , 'mean'), 

    # desv. estandar 
    desv_estandar_minutos=('minutos_jugados', 'std'), 
    desv_estandar_horas=('horas_jugadas', 'std'),
    desv_estandar_goles=('goles', 'std'), 
    desv_estandar_asistencias=('asistencias', 'std'), 
    desv_estandar_tarjestas=('tarjetas' , 'std'), 

).reset_index(drop=False)

type(df_estadisticas_equipo)
```
