```python
import pandas as pd
```

```python
RUTA_RELATIVA = "./excel/Deportes_Examen.xlsx"
```

```python

df = pd.read_excel(RUTA_RELATIVA)
df.head()
```

```python
df.tail()
```

```python
# ver info()
df.info()
```

```python
# Ver nulos (conteo)
df.isnull().sum()
```

```python
# Normalizar los encabezados a minúsculas 
df.columns = df.columns.str.lower()
df.columns
```

```python
df_sort_goles_minuts = df.sort_values(by=["goles", "minutos_jugados"], ascending=False)  # no modifica
df_sort_goles_minuts
```

```python
def MinutesToHour(minutes): 
    return round((minutes/60),2)
```

```python
df["horas_jugadas"] = pd.to_numeric(df['minutos_jugados'], errors='coerce').apply(MinutesToHour)
```

```python
df.head()
```

```python
df['fecha'] = pd.to_datetime(df['fecha'].astype(str), errors='raise', format='%Y-%m-%d')
```

```python
df.info()
```

```python
# generar un DataFrame ordenado por fechas de forma descendente 
df_soted_fecha_desc = df.sort_values(by=['fecha','jugador'], ascending=[False, True])
```

```python
df_soted_fecha_desc.head(10)
```

```python
df_equipos_duplicados = df[df.duplicated(subset=['equipo'], keep=False)].sort_values(by=['equipo', 'jugador'], ascending=True)
```

```python
df_equipos_duplicados.head(15)
```

```python
df_drop_duplicated_equipos = df.drop_duplicates(subset="equipo", keep='first')
```

```python
df.shape
```

```python
df_drop_duplicated_equipos.shape
```

```python
# Obtener nombre del mes
df_sorted_equipos_asc = df.sort_values(by=['equipo', 'goles'], ascending=[True, False])
df_sorted_equipos_asc
```

```python
def DayOfWeek(day): return day + 1 
```

```python
df_sorted_equipos_asc["name_mes"] = df_sorted_equipos_asc['fecha'].dt.month_name()
df_sorted_equipos_asc["name_day"] = df_sorted_equipos_asc['fecha'].dt.day_name()
# como comienza desde 0 le sumamos 1
df_sorted_equipos_asc["idx_day"] = df_sorted_equipos_asc['fecha'].dt.day_of_week.apply(DayOfWeek)
```
