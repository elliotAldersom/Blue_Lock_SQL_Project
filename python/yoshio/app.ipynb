## Atributos y propiedades de pandas.DataFrame

Los atributos y propiedades permiten consultar información sobre la estructura, las dimensiones, los tipos de datos y otras características de un DataFrame. A diferencia de los métodos, generalmente se utilizan sin paréntesis.

| N.º | Atributo / Propiedad | ¿Qué hace? | ¿Qué devuelve? | Ejemplo de uso |
|---:|---|---|---|---|
| 1 | `shape` | Permite conocer las dimensiones del DataFrame, es decir, cuántas filas y columnas contiene. | Una tupla con dos valores: `(filas, columnas)`. | `df.shape` |
| 2 | `columns` | Permite consultar o modificar las etiquetas (nombres) de las columnas del DataFrame. Es una propiedad que dispone de getter y setter. | Un objeto `Index` con los nombres de las columnas. | `df.columns` |
| 3 | `dtypes` | Permite identificar el tipo de dato que pandas reconoce en cada columna. Es fundamental para detectar tipos incorrectos antes de realizar operaciones. | Una Serie de pandas que contiene los tipos de datos de cada columna. | `df.dtypes` |
| 4 | `index` | Permite consultar o modificar las etiquetas que identifican cada fila del DataFrame. Por defecto, pandas utiliza un índice numérico que comienza en 0. | Un objeto `Index` con las etiquetas de las filas. | `df.index` |
| 5 | `size` | Permite conocer la cantidad total de elementos que contiene el DataFrame. Se calcula multiplicando el número de filas por el número de columnas. | Un número entero con el total de elementos. | `df.size` |
| 6 | `empty` | Permite comprobar si el DataFrame está vacío. Devuelve `True` cuando alguna de sus dimensiones tiene longitud cero. | Un valor booleano: `True` o `False`. | `df.empty` |
| 7 | `ndim` | Indica la cantidad de dimensiones de la estructura. Un DataFrame siempre es bidimensional, ya que organiza los datos en filas y columnas. | Un número entero. Para un DataFrame devuelve `2`. | `df.ndim` |
| 8 | `T` | Devuelve la transposición del DataFrame: intercambia las filas por las columnas y viceversa. No modifica el DataFrame original. | Un nuevo DataFrame con los ejes intercambiados. | `df.T` |
| 9 | `values` | Permite acceder a los datos del DataFrame como un array de NumPy, sin conservar las etiquetas de filas y columnas. | Un `numpy.ndarray`. | `df.values` |
| 10 | `axes` | Permite consultar los ejes del DataFrame: el primero representa las filas y el segundo las columnas. | Una lista con dos objetos `Index`: `[index, columns]`. | `df.axes` |
| 11 | `flags` | Permite consultar determinadas características del objeto, como si admite etiquetas duplicadas en sus índices y columnas. | Un objeto `Flags` con indicadores del DataFrame. | `df.flags` |

## Consideraciones importantes

1. **`shape` vs. `size`:** `shape` informa las dimensiones de la estructura, mientras que `size` calcula la cantidad total de elementos.

2. **`columns` vs. `index`:** `columns` identifica las variables (columnas), mientras que `index` identifica las observaciones (filas).

3. **`dtypes`:** cada columna puede tener un tipo de dato diferente, como `int64`, `float64`, `object`, `string` o `datetime64[ns]`.

4. **`empty`:** un DataFrame puede estar vacío aunque tenga etiquetas de columnas. Lo que importa es que alguna de sus dimensiones tenga longitud cero.

5. **`T`:** es útil para reorganizar la visualización de los datos, aunque no suele ser una operación habitual en el análisis de grandes DataFrames.

6. **`values`:** actualmente es preferible utilizar `df.to_numpy()` cuando quieras convertir explícitamente un DataFrame en un array de NumPy, porque permite controlar mejor la conversión.

7. **`flags`:** es una propiedad más especializada. No suele ser necesaria en las primeras etapas de un análisis de datos.

**Documentación oficial:** [pandas.DataFrame — API Reference](https://pandas.pydata.org/docs/reference/frame.html).

### `df.drop(args)` 
- Es un método que permite eliminar etiquetas de filas o columnas de un `DataFrame`. 
**Sintaxix:**
```python
    df.drop(
        labels=None,
        *,
        axis=0,
        index=None,
        columns=None,
        level=None,
        inplace=False,
        errors='raise'
    )
```

**Ejemplo1: con Axis** 
```python 
    # Eliminar la columna 'fecha_corte' con axis
    df.drop('fecha_corte', axis=1) # fecha_corte => 'label'

    # Eliminar el registro con indice 2
    df.drop(2, axis=0) # 2 => 'label'

```
|Valor| Signifcado| Qué elimina| 
|--|---|--|
|`axis=1`| Filas(índices)| Registros
|`axis=0`| Columnas| Variables o campos| 

**Ejemplo2: con `index` y `columns`**
```python 
    # Eliminar la columna 'fecha_corte'
    df.drop(columns=['fecha_corte']) # no necesita axis=1
    
    # Eliminar la fila cuyo indice sea 2 con index
    df.drop(index=2) # no necesita de axis=0
```

**Ejemplo3: `inplace`**
```python 
    # Eliminar la columna 'fecha_corte' modificando el df
    df.drop(columns=['fecha_corte'], inplace=True)
```
|Valor| Significado|
|--|--|
|`inplace=False`|Retornar un nuevo `DataFrame` con los cambios aplicados, es el valor predeterminado| 
|`inplace=True`|Aplicar los cambios en el `DataFrame` original|  

**Ejemplo4: `error`** 
```python 
    # Si 'fecha_corte' no existe, pandas genera un KeyError
    df.drop(columns=['fecha_corte'], errors='raise')

    # Si no existe, simplemente la ignora
    df.drop(columns=['fecha_corte'], errors='ignore')
``` 
|valor|Significado|
|---|---|
|`errors='raise'`|Genera un `KeyError` si la etiqueta no existe. Valor por defecto.|
|`errors='ignore'`|Ignora las etiquetas que no encuentra y continua| 

````python
# df.drop(index=[20000000000]).head() # => KeyError 
df.drop(index=[20000000000], errors='ignore').head() # => Ignora el error y continua.
````

### `df.astype(args)`
- Método de `pandas` que permite convertir el tipo de dato `(dtype)` de una `Serie` o de las columnas de un `DataFrame`.

**Sintaxis:**
```python 
objeto.astype(dtype, copy=None, errors='raise')
```

|Argumentos|Función|
|--|--|
|`dtype`|Tipo de dato al que deseas convertir, es obligatorio.|
|`copy`|Controla si se solicita una copia del objeto.|
|`errors`|Determina que hacer si la conversión falla.| 

#### dtype
**Ejemplo:**
```python 
    import pandas as pd

    df = pd.DataFrame({
        'edad': ['19', '23', '25'], # str 
        'precio': [10.5, 20.5, 30.5],
        'codigo': [101, 102, 103]
    })

    print(df.dtypes) # atributo para ver el tipo de cada columna
    # edad       object   => str aunque lo muestre como object
    # precio    float64
    # codigo      int64
    # dtype: object

    df['edad'] = df['edad'].astype(int) # cambiamos de str => int

    print(df.dtypes)
    # edad       int64
    # precio    float64
    # codigo      int64
    # dtype: object
``` 

**Ejemplo: Convertir más de una columna**
```python 
    # Usas un diccionario donde: 
    # 'key' => nombre columna 
    # valor 'value' => tipo de dato a convertir
    df = df.astype({
        'edad': 'int64',
        'precio': 'float64',
        'codigo': 'str'
    })

    df = df.astype(str) # convierte todas las columnas a 'str'
``` 
#### copy => valor por defecto None
**Ejemplo:**

```python 
df['edad'].astype('float64', copy=True)
```
|Valor|Significado|
|--|--|
|`True`|Solicita una copia| 
| `False`|Permite evitar una copia cuando sea posible|
|`None`| Deja que pandas determine el comportamiento|

> Ahora no es necesario controlar la copia manualmente, así que este argumento no se recomienda modificar y es mejor dejar en `None`.

#### errors => por defecto 'raise'
|Valor|Significado|
|--|--|
|`'raise'`|Geenra un error si encuentra un valor que no pueda convertir|
|`'ignore'`|Devuelve el objeto original si la conversión falla. No se recomienda usar este valor| 

**Como ni `copy` y ni `errors` se recomiendan modificar, solo debes de pasar comoa argumento el `dtype`.**

### `pd.to_numeric(args)` / `pandas.to_numeric(args)`
- Es una función de `Pandas` que convierte valores a un tipo de dato numérico. 
- Se utiliza principalmente cuando importamos datos de archivos `CSV` o `Excel (xlsx)` u otras fuentes y encontramos columnas que contienen números almacenados como `texto`, o valores que no se pueden interpretat directamente como números.
- A diferencia del método `astype()`, que permite convertir a distintos tipos de datos, `pd.to_numeric()` está especializada en la conversión de tipos `numéricos`. 

**Sintaxis:** 
```python 
    pd.to_numeric(arg, arrors, downcast)
```

|Parámetro|Explicación|
|--|--|
|`arg`|Objeto cuyos valores queremos convertir a números, puede ser una columna de un `DataFrame`, entre otros objetos.|
|`errors`|Determina qué sucede cuando `pandas` encuentra un valor que no se puede convertir a número.|
|`downcast`|Permite reducir el tamaño del tipo numérico cuando es posible, por ejemplo, de `int64` a `int8`|

**OJO:** `pd.to_numeric(args)` devuelve los valores convertios, pero no modifica el `DataFrame` original, por eso lo reasignamos con `df['column_name']` 

#### arg

**Ejemplo:**
```python
    import pandas as pd

    df = pd.DataFrame({
        'nombre': ['Elliot', 'Darlene', 'Angela', 'Tyrell'],
        'edad': ['19', '25', '23', '30'],
        'salario': ['1500.50', '2000.00', '1800.75', '2500.00']
    })

    df.dtypes

    # nombre     object
    # edad       object
    # salario    object
    # dtype: object

    df['edad'] = pd.to_numeric(df['edad'])
    df['salario'] = pd.to_numeric(df['salario'])    

    df.dtypes

    # nombre      object
    # edad         int64
    # salario    float64
    # dtype: object
```

#### errors => por defecto 'raise'
|Valor|Qué hace| 
|-|-|
|`'raise'`|Si un valor no puede convertirse, lanzará un error.| 
|`'coerce'`|Si un valor no se puede convertir, lo transforma en `NaN` `(Not a Number)`| 
|`'ignore'`|Si un valor no se puede convertir, devuelve el objeto sin realizar la conversión.|

> Acerca del `downcast` no es necesario ponerlo.


````python
df['latitud'] = pd.to_numeric(df['latitud'], errors='coerce') # 'Hola' -> NaN
df['longitud'] = pd.to_numeric(df['longitud'], errors='coerce') # 'Hola' -> NaN
df['profundidad'] = pd.to_numeric(df['profundidad'], errors='coerce') # 'Hola' -> NaN
df['magnitud'] = pd.to_numeric(df['magnitud'], errors='coerce') # 'Hola' -> NaN
````

````python
df.isnull().sum()  # no se encontró nulos al cambiar al tipo de dato de cada columna
````

### `pd.to_datetime(arg)` / `pandas.to_datetime(arg)`

- Es una función de `Pandas` que convierte valores a un tipo de dato de fecha y hora (`datetime`).
- Se utiliza principalmente cuando importamos datos de archivos `CSV`, `Excel (xlsx)` u otras fuentes y encontramos columnas que contienen fechas almacenadas como `texto` (`object` o `string`).
- Permite interpretar diferentes formatos de fechas y convertirlos a un formato temporal que `Pandas` pueda reconocer, comparar, ordenar y utilizar en operaciones relacionadas con fechas.
- A diferencia del método `astype()`, que permite convertir a distintos tipos de datos, `pd.to_datetime()` está especializada en la conversión de valores temporales (tiempo).

**Sintaxis:**
```python
pd.to_datetime(
    arg,
    errors='raise',
    dayfirst=False,
    yearfirst=False,
    utc=False,
    format=None,
    exact=True,
    unit=None,
    origin='unix',
    cache=True
)
```

| Parámetro | Explicación |
|--|--|
| `arg` | Objeto cuyos valores queremos convertir a fechas y horas. Puede ser una columna de un `DataFrame`, entre otros objetos. |
| `errors` | Determina qué sucede cuando `Pandas` encuentra un valor que no puede interpretar como fecha. |
| `dayfirst` | Si es `True`, intenta interpretar primero el día en fechas numéricas ambiguas, por ejemplo, `15/02/2026`. |
| `yearfirst` | Si es `True`, intenta interpretar primero el año en fechas numéricas ambiguas. |
| `utc` | Si es `True`, convierte las fechas a una representación con zona horaria UTC. |
| `format` | Permite especificar el formato exacto o esperado de las fechas que queremos interpretar. |
| `exact` | Determina si el formato indicado debe coincidir exactamente con la cadena de fecha. |
| `unit` | Especifica la unidad de tiempo cuando los valores son marcas de tiempo numéricas (por ejemplo, `'s'`, `'ms'`, `'us'`, `'ns'`). |
| `origin` | Define el punto de origen desde el cual se calculan las marcas de tiempo numéricas. Por defecto es `'unix'`. |
| `cache` | Permite almacenar temporalmente resultados de conversión repetidos para intentar acelerar el proceso. |

**OJO:** `pd.to_datetime()` devuelve los valores convertidos, pero no modifica el `DataFrame` original. Por eso lo reasignamos utilizando `df['column_name']`.

Además, el resultado normalmente tendrá un tipo de dato `datetime64[ns]` cuando trabajamos con fechas sin zona horaria.

#### arg

**Ejemplo:**

```python
import pandas as pd

df = pd.DataFrame({
    'nombre': ['Elliot', 'Darlene', 'Angela', 'Tyrell'],
    'fecha_nacimiento': [
        '15/03/2005',
        '21/07/2001',
        '10/11/2003',
        '05/01/1999'
    ]
})

df.dtypes

# nombre              object
# fecha_nacimiento    object
# dtype: object

df['fecha_nacimiento'] = pd.to_datetime(
    df['fecha_nacimiento'],
    format='%d/%m/%Y'
)

df.dtypes

# nombre                      object
# fecha_nacimiento    datetime64[ns]
# dtype: object
```

Observa que `fecha_nacimiento` pasó de `object` a `datetime64[ns]`.

Esto significa que `Pandas` ya reconoce los valores como fechas y no simplemente como cadenas de texto.

#### errors => por defecto 'raise'

| Valor | Qué hace |
|--|--|
| `'raise'` | Si un valor no puede interpretarse como fecha, lanza un error (`ValueError`). |
| `'coerce'` | Si un valor no se puede interpretar como fecha, lo transforma en `NaT` (`Not a Time`), que representa un valor temporal ausente o inválido. |
| `'ignore'` | Opción obsoleta (`deprecated`) en versiones modernas de Pandas. No se recomienda utilizarla. |

**Ejemplo con `errors='coerce'`:**

```python
df = pd.DataFrame({
    'nombre': ['Elliot', 'Darlene', 'Angela'],
    'fecha': ['15/03/2005', 'fecha inválida', '10/11/2003']
})

df['fecha'] = pd.to_datetime(
    df['fecha'],
    format='%d/%m/%Y',
    errors='coerce'
)

# Resultado:
#
#     nombre       fecha
# 0   Elliot       2005-03-15
# 1   Darlene      NaT
# 2   Angela       2003-11-10
```

**Importante:** `NaT` significa `Not a Time`. Es el equivalente temporal de `NaN`, pero para datos de fecha y hora.

Podemos identificar estos valores con los mismos métodos de pandas que utilizamos para detectar valores ausentes:

```python
df.isnull().sum()
```

#### dayfirst y yearfirst

Estos parámetros permiten indicar cómo queremos interpretar fechas numéricas ambiguas.

| Parámetro | Valor | Interpretación preferida |
|--|--|--|
| `dayfirst` | `True` | Día / mes / año |
| `dayfirst` | `False` | Mes / día / año (comportamiento predeterminado) |
| `yearfirst` | `True` | Año / mes / día |
| `yearfirst` | `False` | No prioriza el año al interpretar fechas ambiguas. |

**Ejemplo:**

```python
df = pd.DataFrame({
    'fecha': ['03/04/2026', '05/06/2026']
})

# Priorizar día / mes / año
df['fecha'] = pd.to_datetime(
    df['fecha'],
    dayfirst=True
)
```

En este caso, `03/04/2026` se interpreta preferentemente como 3 de abril de 2026.

**OJO:** `dayfirst` y `yearfirst` son preferencias de interpretación, no validaciones estrictas del formato. Si conocemos el formato de las fechas, es preferible especificarlo mediante `format`.

#### format

Este parámetro permite indicar explícitamente cómo están escritas las fechas para que `Pandas` pueda interpretarlas correctamente.

Algunos códigos frecuentes:

| Código | Significado | Ejemplo |
|--|--|--|
| `%d` | Día del mes | `15` |
| `%m` | Mes numérico | `03` |
| `%Y` | Año con cuatro dígitos | `2026` |
| `%y` | Año con dos dígitos | `26` |
| `%H` | Hora (formato 24 horas) | `14` |
| `%M` | Minutos | `35` |
| `%S` | Segundos | `50` |

**Ejemplos de formatos:**

| Fecha original | `format` |
|--|--|
| `15/03/2026` | `'%d/%m/%Y'` |
| `2026-03-15` | `'%Y-%m-%d'` |
| `03-15-2026` | `'%m-%d-%Y'` |
| `15/03/2026 14:35:50` | `'%d/%m/%Y %H:%M:%S'` |

**Ejemplo:**

```python
df = pd.DataFrame({
    'fecha': ['15/03/2026', '21/07/2026']
})

df['fecha'] = pd.to_datetime(
    df['fecha'],
    format='%d/%m/%Y'
)
```

Aquí le indicamos a `Pandas` que las fechas tienen el formato día/mes/año.

#### unit y origin

Estos parámetros se utilizan principalmente cuando tenemos marcas de tiempo numéricas (`timestamps`), es decir, números que representan una cantidad de tiempo transcurrido desde un punto de origen.

- `unit`: indica la unidad de tiempo de los números.
- `origin`: indica desde qué fecha comienza el cálculo.

Unidades frecuentes:

| Valor | Significado |
|--|--|
| `'D'` | Días |
| `'s'` | Segundos |
| `'ms'` | Milisegundos |
| `'us'` | Microsegundos |
| `'ns'` | Nanosegundos |

**Ejemplo:**

```python
df = pd.DataFrame({
    'tiempo': [0, 86400, 172800]
})

df['fecha'] = pd.to_datetime(
    df['tiempo'],
    unit='s',
    origin='unix'
)
```

Aquí los números representan segundos transcurridos desde el origen Unix (`1970-01-01 00:00:00 UTC`).

Por tanto, cada `86400` segundos representa un día.

#### utc

Este parámetro permite trabajar con fechas y horas que tienen como referencia la zona horaria UTC.

```python
df = pd.DataFrame({
    'fecha': ['2026-03-15 14:30:00']
})

df['fecha'] = pd.to_datetime(
    df['fecha'],
    utc=True
)
```

El resultado tendrá información de zona horaria (`UTC`), a diferencia de una fecha sin zona horaria.

**OJO:** `utc=True` no significa simplemente agregar la etiqueta UTC. Si la fecha ya tiene una zona horaria, Pandas convierte el instante correspondiente a UTC. Si no tiene zona horaria, interpreta el valor como si ya estuviera expresado en UTC.

#### Acerca de los parámetros

No es necesario especificar todos los parámetros de `pd.to_datetime()`.

- `errors`: conviene conocerlo para controlar fechas inválidas.


### `DataFrame.describe()` / `df.describe()`

- Es un método de `Pandas` que genera un resumen estadístico descriptivo de las columnas de un `DataFrame`.
- Se utiliza principalmente durante el análisis exploratorio de datos (EDA), para conocer rápidamente cómo se distribuyen los valores de nuestras columnas y detectar posibles anomalías.
- Calcula diferentes medidas estadísticas, como la cantidad de valores presentes (`count`), la media (`mean`), la desviación estándar (`std`), el valor mínimo (`min`), los cuartiles (`25%`, `50%`, `75%`) y el valor máximo (`max`).
- Por defecto, `describe()` analiza las columnas numéricas del `DataFrame`. También puede analizar columnas de texto, fechas y otros tipos de datos mediante el parámetro `include`.

**Sintaxis:**

```python
df.describe(
    percentiles=None,
    include=None,
    exclude=None,
    datetime_is_numeric=False
)
```

**Nota:** La firma puede variar según la versión de Pandas. En versiones actuales, `datetime_is_numeric` ya no forma parte de la API de `DataFrame.describe()`.

| Parámetro | Explicación |
|--|--|
| `percentiles` | Permite especificar qué percentiles queremos calcular, por ejemplo, `[0.10, 0.50, 0.90]`. |
| `include` | Permite especificar qué tipos de columnas queremos incluir en el análisis. |
| `exclude` | Permite excluir determinados tipos de columnas del análisis. |

**OJO:** `describe()` devuelve un nuevo `DataFrame` con el resumen estadístico. No modifica el `DataFrame` original.

#### Ejemplo

```python
import pandas as pd

df = pd.DataFrame({
    'nombre': ['Elliot', 'Darlene', 'Angela', 'Tyrell'],
    'edad': [19, 25, 23, 30],
    'salario': [1500, 2000, 1800, 2500]
})

df.describe()
```

**Resultado:**

| Estadístico | edad | salario |
|--|--:|--:|
| count | 4.00 | 4.00 |
| mean | 24.25 | 1950.00 |
| std | 4.65 | 416.33 |
| min | 19.00 | 1500.00 |
| 25% | 22.00 | 1725.00 |
| 50% | 24.00 | 1900.00 |
| 75% | 26.25 | 2125.00 |
| max | 30.00 | 2500.00 |

#### ¿Qué significa cada estadístico?

| Estadístico | Explicación |
|--|--|
| `count` | Cantidad de valores no ausentes que existen en la columna. No cuenta los valores `NaN`. |
| `mean` | Media aritmética: suma todos los valores y divide el resultado entre la cantidad de valores presentes. |
| `std` | Desviación estándar: mide cuánto se dispersan los valores respecto a la media. Cuanto mayor sea, mayor será la dispersión. |
| `min` | Valor mínimo encontrado en la columna. |
| `25%` | Primer cuartil (Q1): indica que el 25% de los valores se encuentra por debajo o igual a este valor. |
| `50%` | Segundo cuartil (Q2) o mediana: divide los datos ordenados en dos mitades. El 50% está por debajo o igual a este valor. |
| `75%` | Tercer cuartil (Q3): indica que el 75% de los valores se encuentra por debajo o igual a este valor. |
| `max` | Valor máximo encontrado en la columna. |

**IMPORTANTE:** Los percentiles y cuartiles no representan promedios. Son medidas que permiten conocer la posición de los valores dentro de un conjunto de datos.

#### percentiles

Este parámetro permite indicar qué percentiles adicionales queremos calcular.

Por defecto, `describe()` utiliza los percentiles:

- `0.25` → 25%.
- `0.50` → 50%.
- `0.75` → 75%.

**Ejemplo:**

```python
df.describe(
    percentiles=[0.10, 0.25, 0.50, 0.75, 0.90]
)
```

Ahora tendremos también:

- `10%`: valor por debajo o igual al cual se encuentra el 10% de los datos.
- `90%`: valor por debajo o igual al cual se encuentra el 90% de los datos.

Podemos agregar los percentiles que necesitemos, siempre que estén entre `0` y `1` y no sean duplicados.

#### include

Por defecto, `describe()` analiza las columnas numéricas.

Con `include` podemos indicar qué tipos de datos queremos incluir en el resumen.

| Valor | Qué incluye |
|--|--|
| `None` | Comportamiento predeterminado: analiza las columnas numéricas. |
| `'all'` | Incluye todas las columnas, independientemente de su tipo. |
| `'number'` | Incluye todas las columnas numéricas. |
| `'object'` | Incluye las columnas de tipo `object`, normalmente utilizadas para texto. |
| `'int64'` | Incluye las columnas de tipo entero `int64`. |
| `'float64'` | Incluye las columnas de tipo decimal `float64`. |

**Ejemplo:**

```python
# Analizar únicamente columnas numéricas
df.describe(include='number')

# Analizar únicamente columnas de texto
df.describe(include='object')

# Analizar todas las columnas
df.describe(include='all')
```

**IMPORTANTE:** Cuando analizamos columnas de texto, `describe()` devuelve estadísticas diferentes:

| Estadístico | Explicación |
|--|--|
| `count` | Cantidad de valores no ausentes. |
| `unique` | Cantidad de valores únicos diferentes. |
| `top` | Valor que aparece con mayor frecuencia (moda). |
| `freq` | Cantidad de veces que aparece el valor más frecuente. |

**Ejemplo:**

```python
df.describe(include='object')
```

Si tenemos una columna `nombre` con los valores:

```text
Elliot
Darlene
Angela
Elliot
```

El resumen será:

| Estadístico | nombre |
|--|--|
| count | 4 |
| unique | 3 |
| top | Elliot |
| freq | 2 |

Observa que `Elliot` aparece dos veces, por eso es el valor más frecuente (`top`).

#### exclude

Este parámetro permite excluir determinados tipos de columnas del análisis.

**Ejemplo:**

```python
# Excluir columnas de texto
df.describe(exclude='object')
```

También podemos excluir otros tipos de datos:

```python
df.describe(exclude=['object', 'bool'])
```

Esto resulta útil cuando queremos analizar todos los tipos de columnas excepto algunos específicos.

#### Acerca de los parámetros

No es necesario especificar todos los parámetros de `describe()`.

- `percentiles`: se utiliza cuando necesitamos analizar posiciones específicas de la distribución de los datos.
- `include`: se utiliza cuando queremos analizar columnas de texto, fechas u otros tipos de datos, además de las numéricas.
- `exclude`: se utiliza cuando queremos omitir determinados tipos de columnas.
- Por defecto, podemos ejecutar simplemente `df.describe()` y Pandas generará el resumen estadístico de las columnas numéricas.

**En resumen:** `describe()` es una herramienta fundamental del análisis exploratorio de datos (EDA), porque permite conocer rápidamente la distribución, dispersión y valores extremos de nuestras variables numéricas, así como obtener resúmenes de las columnas categóricas.


## `DataFrame.duplicated()`

El método `duplicated()` permite **identificar las filas duplicadas de un DataFrame**. Devuelve valores booleanos (`True` o `False`) para indicar si una fila ya apareció anteriormente, considerando todas las columnas o las columnas que especifiquemos.

Es especialmente útil durante la limpieza de datos (Data Cleaning), porque permite detectar registros repetidos antes de eliminarlos.

### Sintaxis

```python
df.duplicated(
    subset=None,
    keep='first'
)
```

### Parámetros

| Parámetro | Descripción |
|---|---|
| `subset` | Permite especificar qué columnas se utilizarán para comprobar si existen duplicados. Si se deja en `None`, se comparan todas las columnas del DataFrame. |
| `keep` | Determina qué aparición de un registro duplicado se conservará como `False` y cuáles se marcarán como `True`. |

### Funcionamiento de `keep`

El parámetro `keep` tiene tres opciones:

| Valor | Comportamiento |
|---|---|
| `'first'` | Es el valor predeterminado. Conserva la primera aparición como `False` y marca las siguientes apariciones como `True`. |
| `'last'` | Conserva la última aparición como `False` y marca las anteriores como `True`. |
| `False` | Marca todas las apariciones de un registro duplicado como `True`, incluida la primera. |

### Ejemplo

```python
import pandas as pd

df = pd.DataFrame({
    'nombre': ['Elliot', 'Darlene', 'Angela', 'Elliot', 'Tyrell', 'Darlene'],
    'edad': [19, 25, 23, 19, 30, 25]
})

print(df.duplicated())
```

Resultado:

```text
0    False
1    False
2    False
3     True
4    False
5     True
dtype: bool
```

**¿Por qué ocurre esto?**

- La fila 0 (`Elliot`, 19) aparece por primera vez, por eso devuelve `False`.
- La fila 3 contiene los mismos valores que la fila 0, por eso devuelve `True`.
- La fila 1 (`Darlene`, 25) aparece por primera vez, por eso devuelve `False`.
- La fila 5 repite los valores de la fila 1, por eso devuelve `True`.

### Parámetro `subset`

Permite comprobar duplicados utilizando únicamente las columnas que indiquemos, en lugar de comparar todas las columnas del DataFrame.

```python
df.duplicated(subset=['nombre'])
```

En este caso, pandas solamente compara los valores de la columna `nombre`.

Esto significa que, aunque dos registros tengan edades diferentes, serán considerados duplicados si tienen el mismo nombre.

También podemos especificar varias columnas:

```python
df.duplicated(subset=['nombre', 'edad'])
```

Aquí pandas considera duplicados los registros que tengan el mismo nombre y la misma edad.

### ¿Qué ocurre al utilizar `duplicated().sum()`?

Es una forma práctica de **contar cuántas filas duplicadas existen en un DataFrame**.

```python
df.duplicated().sum()
```

¿Por qué funciona?

1. `duplicated()` devuelve una serie de valores booleanos.
2. En Python, los valores booleanos pueden interpretarse numéricamente:
   - `True` equivale a `1`.
   - `False` equivale a `0`.
3. `sum()` suma esos valores, por lo que cuenta cuántos `True` existen.

Por ejemplo:

```python
df.duplicated().sum()
```

Resultado:

```text
2
```

Esto significa que existen **2 filas duplicadas**, considerando todas las columnas y utilizando el comportamiento predeterminado (`keep='first'`).

También podemos combinarlo con `subset`:

```python
df.duplicated(subset=['nombre']).sum()
```

Esto cuenta cuántas filas están duplicadas según la columna `nombre`.

**Importante:** con `keep='first'`, el conteo no incluye la primera aparición de cada registro repetido. Si queremos contar todas las filas pertenecientes a grupos duplicados, debemos utilizar:

```python
df.duplicated(keep=False).sum()
```

### Diferencia entre `duplicated()` y `drop_duplicates()`

Aunque ambos métodos trabajan con registros duplicados, tienen propósitos diferentes:

| Método | Función |
|---|---|
| `duplicated()` | Identifica las filas duplicadas y devuelve valores booleanos (`True` o `False`). No elimina registros. |
| `drop_duplicates()` | Devuelve un DataFrame sin las filas duplicadas, conservando las apariciones que indique `keep`. |

Ejemplo:

```python
# Identificar duplicados
df.duplicated()

# Contar duplicados
df.duplicated().sum()

# Eliminar duplicados
df = df.drop_duplicates()
```

**En resumen:** `duplicated()` permite detectar registros repetidos, mientras que `duplicated().sum()` permite contar cuántos registros están duplicados. Si necesitamos eliminarlos, utilizamos `drop_duplicates()`.


````python
df.duplicated().sum() # sumar todos los registros duplicados
````

````python
df.duplicated(subset=['fecha_utc', 'hora_utc', 'latitud', 'longitud'])
````

````python
# indexación booleana
duplicados = df[df.duplicated(subset=['fecha_utc', 'hora_utc', 'latitud', 'longitud'])]
duplicados
````


## `DataFrame.sort_values()`

El método `sort_values()` permite **ordenar las filas de un DataFrame según los valores de una o varias columnas**.

Se utiliza para organizar los registros de acuerdo con un criterio específico, por ejemplo, ordenar personas por edad, productos por precio o ventas de mayor a menor.

### Sintaxis

```python
df.sort_values(
    by,
    axis=0,
    ascending=True,
    inplace=False,
    kind='quicksort',
    na_position='last',
    ignore_index=False,
    key=None
)
```

### Parámetros

| Parámetro | Descripción |
|---|---|
| `by` | Indica la columna o las columnas que se utilizarán para ordenar el DataFrame. Es obligatorio. |
| `axis` | Determina el eje sobre el cual se realizará el ordenamiento. `0` ordena las filas según los valores de las columnas y `1` ordena las columnas según los valores de las filas. Normalmente utilizaremos `0`. |
| `ascending` | Determina el sentido del ordenamiento. `True` ordena de menor a mayor (ascendente) y `False` de mayor a menor (descendente). |
| `inplace` | Si es `True`, modifica directamente el DataFrame original. Si es `False`, devuelve un nuevo DataFrame ordenado sin modificar el original. |
| `kind` | Especifica el algoritmo de ordenamiento. Algunas opciones son `'quicksort'`, `'mergesort'`, `'heapsort'` y `'stable'`. |
| `na_position` | Determina dónde se ubicarán los valores faltantes (`NaN`). Puede ser `'first'` para colocarlos al inicio o `'last'` para colocarlos al final. |
| `ignore_index` | Si es `True`, el DataFrame resultante tendrá un índice nuevo y consecutivo (`0, 1, 2...`). Si es `False`, conserva las etiquetas del índice original. |
| `key` | Permite aplicar una función a los valores de la columna antes de ordenarlos. Por ejemplo, se puede utilizar `str.lower` para ordenar textos sin distinguir entre mayúsculas y minúsculas. |

### Parámetro `by`

Es el parámetro obligatorio que indica según qué columna o columnas queremos ordenar el DataFrame.

**Podemos especificar una sola columna:**

```python
df.sort_values(by='edad')
```

**O varias columnas:**
```python
df.sort_values(by=['edad', 'nombre'])
```

Cuando indicamos varias columnas, pandas utiliza la primera como criterio principal. Si encuentra valores iguales en esa columna, utiliza la siguiente para decidir el orden.

### Parámetro `ascending`

Determina si el ordenamiento será ascendente o descendente.

```python
# Orden ascendente: de menor a mayor
df.sort_values(by='edad', ascending=True)

# Orden descendente: de mayor a menor
df.sort_values(by='edad', ascending=False)
```

`ascending=True` es el valor predeterminado, por lo que podemos omitirlo cuando queremos ordenar de menor a mayor.

También podemos especificar un sentido diferente para cada columna:

```python
df.sort_values(
    by=['edad', 'salario'],
    ascending=[True, False]
)
```

**En este caso:**
- Primero ordena por `edad`, de menor a mayor.
- Cuando encuentra personas con la misma edad, ordena sus salarios de mayor a menor.

### Ejemplo

```python
import pandas as pd

df = pd.DataFrame({
    'nombre': ['Elliot', 'Darlene', 'Angela', 'Tyrell'],
    'edad': [25, 19, 30, 25],
    'salario': [2000, 1500, 3000, 2500]
})

df_ordenado = df.sort_values(by='edad')

print(df_ordenado)
```

**Resultado:**

```text
    nombre  edad  salario
1  Darlene    19     1500
0   Elliot    25     2000
3   Tyrell    25     2500
2   Angela    30     3000
```

**¿Por qué los índices aparecen desordenados?**

Porque `sort_values()` reorganiza las filas, pero conserva sus índices originales por defecto.

Por ejemplo, Darlene originalmente tenía el índice `1`. Al ordenar el DataFrame, su fila cambia de posición, pero sigue conservando ese índice.

Si queremos que el resultado tenga índices consecutivos, podemos utilizar:

```python
df_ordenado = df.sort_values(
    by='edad',
    ignore_index=True
)
```

**Resultado:**

```text
    nombre  edad  salario
0  Darlene    19     1500
1   Elliot    25     2000
2   Tyrell    25     2500
3   Angela    30     3000
```

### Parámetro `inplace`

Por defecto, `inplace=False`, lo que significa que `sort_values()` devuelve un nuevo DataFrame ordenado y no modifica el original.

```python
df_ordenado = df.sort_values(by='edad')
```

Si utilizamos `inplace=True`, el ordenamiento se aplica directamente sobre el DataFrame original:

```python
df.sort_values(by='edad', inplace=True)
```

En este caso, no necesitamos reasignar el resultado a `df`, porque el propio DataFrame ya fue modificado.

**Importante:** cuando utilizamos `inplace=True`, el método devuelve `None`, no un DataFrame. Por eso no debemos hacer lo siguiente:

```python
df = df.sort_values(by='edad', inplace=True)
```

Esto provocaría que `df` termine teniendo el valor `None`.

### Parámetro `na_position`

Permite decidir dónde se colocarán los valores faltantes (`NaN`) durante el ordenamiento.

```python
# Colocar NaN al final
df.sort_values(by='edad', na_position='last')

# Colocar NaN al principio
df.sort_values(by='edad', na_position='first')
```

Por defecto, utiliza `'last'`, por lo que los valores faltantes se colocan al final.

### Parámetro `key`

Permite transformar temporalmente los valores de una columna antes de ordenarlos.

**Por ejemplo:**

```python
df.sort_values(
    by='nombre',
    key=lambda columna: columna.str.lower()
)
```

Esto permite ordenar los nombres sin distinguir entre mayúsculas y minúsculas.

La transformación se utiliza únicamente para determinar el orden; no modifica los valores originales de la columna.

### Diferencia entre `sort_values()` y `sort_index()`

Aunque ambos métodos permiten ordenar un DataFrame, utilizan criterios diferentes:

| Método | Función |
|---|---|
| `sort_values()` | Ordena las filas según los valores contenidos en una o varias columnas. |
| `sort_index()` | Ordena las filas o columnas según las etiquetas de su índice. |

**Ejemplo:**

```python
# Ordenar según los valores de la columna edad
df.sort_values(by='edad')

# Ordenar según las etiquetas del índice
df.sort_index()
```

**En resumen:** `sort_values()` permite ordenar los registros de un DataFrame según los valores de sus columnas. Sus parámetros más utilizados son `by`, `ascending`, `inplace`, `na_position` e `ignore_index`.



````python
duplicados.sort_values(by=['fecha_utc', 'hora_utc'])
````


## `DataFrame.drop_duplicates()`

El método `drop_duplicates()` permite **eliminar las filas duplicadas de un DataFrame**, conservando una sola aparición de cada registro repetido, según los criterios que especifiquemos.

Se utiliza principalmente durante la limpieza de datos (*Data Cleaning*), cuando necesitamos eliminar registros repetidos para evitar duplicaciones en nuestros análisis.

### Sintaxis

```python
df.drop_duplicates(
    subset=None,
    keep='first',
    inplace=False,
    ignore_index=False
)
```

### Parámetros

| Parámetro | Descripción |
|---|---|
| `subset` | Permite especificar qué columnas se utilizarán para identificar los registros duplicados. Si se deja en `None`, se comparan todas las columnas del DataFrame. |
| `keep` | Determina qué aparición de un registro duplicado se conservará. Puede ser `'first'`, `'last'` o `False`. |
| `inplace` | Si es `True`, modifica directamente el DataFrame original. Si es `False`, devuelve un nuevo DataFrame sin duplicados, sin modificar el original. |
| `ignore_index` | Si es `True`, el DataFrame resultante tendrá un índice nuevo y consecutivo (`0, 1, 2...`). Si es `False`, conserva las etiquetas del índice original. |

### Funcionamiento de `keep`

El parámetro `keep` determina qué aparición de un registro duplicado queremos conservar.

Tiene tres opciones:

| Valor | Comportamiento |
|---|---|
| `'first'` | Es el valor predeterminado. Conserva la primera aparición de un registro y elimina las siguientes apariciones duplicadas. |
| `'last'` | Conserva la última aparición de un registro y elimina las anteriores apariciones duplicadas. |
| `False` | Elimina todas las apariciones de los registros duplicados, incluida la primera. |

### Ejemplo

```python
import pandas as pd

df = pd.DataFrame({
    'nombre': ['Elliot', 'Darlene', 'Angela', 'Elliot', 'Tyrell', 'Darlene'],
    'edad': [19, 25, 23, 19, 30, 25]
})

df_sin_duplicados = df.drop_duplicates()

print(df_sin_duplicados)
```

Resultado:

```text
    nombre  edad
0   Elliot    19
1  Darlene    25
2   Angela    23
4   Tyrell    30
```

**¿Por qué ocurre esto?**

- La fila 0 (`Elliot`, 19) se conserva porque es la primera aparición.
- La fila 3 contiene los mismos valores que la fila 0, por eso se elimina.
- La fila 1 (`Darlene`, 25) se conserva porque es la primera aparición.
- La fila 5 repite los valores de la fila 1, por eso se elimina.

Por defecto, pandas compara todas las columnas y conserva la primera aparición de cada registro.

### Parámetro `subset`

Permite identificar duplicados utilizando únicamente las columnas que indiquemos, en lugar de comparar todas las columnas del DataFrame.

```python
df_sin_duplicados = df.drop_duplicates(subset=['nombre'])
```

En este caso, pandas solamente compara los valores de la columna `nombre`.

Esto significa que, aunque dos registros tengan edades diferentes, serán considerados duplicados si tienen el mismo nombre. Por lo tanto, se conservará una sola aparición de cada nombre.

También podemos especificar varias columnas:

```python
df_sin_duplicados = df.drop_duplicates(
    subset=['nombre', 'edad']
)
```

Aquí pandas considera duplicados los registros que tengan el mismo nombre y la misma edad.

**Importante:** las columnas indicadas en `subset` se utilizan para determinar qué registros son duplicados, pero no se eliminan las demás columnas del DataFrame. El resultado conserva todas las columnas originales.

### Parámetro `inplace`

Por defecto, `inplace=False`, lo que significa que `drop_duplicates()` devuelve un nuevo DataFrame sin duplicados y no modifica el original.

```python
df_sin_duplicados = df.drop_duplicates()
```

Si utilizamos `inplace=True`, la eliminación se aplica directamente sobre el DataFrame original:

```python
df.drop_duplicates(inplace=True)
```

En este caso, no necesitamos reasignar el resultado a `df`, porque el propio DataFrame ya fue modificado.

**Importante:** cuando utilizamos `inplace=True`, el método devuelve `None`, no un DataFrame. Por eso no debemos hacer lo siguiente:

```python
df = df.drop_duplicates(inplace=True)
```

Esto provocaría que `df` termine teniendo el valor `None`.

### Parámetro `ignore_index`

Permite decidir si queremos conservar los índices originales o generar un nuevo índice consecutivo después de eliminar los duplicados.

```python
df_sin_duplicados = df.drop_duplicates(
    ignore_index=True
)
```

Por ejemplo, si originalmente tenemos los índices `0, 1, 2, 3, 4, 5` y se eliminan las filas duplicadas con índices `3` y `5`, sin `ignore_index=True` el resultado conservará los índices `0, 1, 2, 4`.

Con `ignore_index=True`, pandas genera nuevamente los índices `0, 1, 2, 3`.

### Diferencia entre `duplicated()` y `drop_duplicates()`

Aunque ambos métodos trabajan con registros duplicados, tienen propósitos diferentes:

| Método | Función |
|---|---|
| `duplicated()` | Identifica las filas duplicadas y devuelve valores booleanos (`True` o `False`). No elimina registros. |
| `drop_duplicates()` | Devuelve un DataFrame sin las filas duplicadas, conservando las apariciones que indique `keep`. |

Ejemplo:

```python
# Identificar las filas duplicadas
df.duplicated()

# Contar cuántas filas están duplicadas
df.duplicated().sum()

# Eliminar las filas duplicadas
df = df.drop_duplicates()
```

**En resumen:** `duplicated()` permite detectar registros repetidos, mientras que `drop_duplicates()` permite eliminarlos. Ambos métodos admiten `subset` para determinar qué columnas se utilizarán como criterio de comparación.


## `DataFrame.apply()`

El método `apply()` permite **aplicar una función a las filas o columnas de un DataFrame**.

Se utiliza cuando necesitamos realizar operaciones personalizadas sobre los datos, por ejemplo, calcular un resultado a partir de varias columnas, transformar valores o aplicar una función que hayamos creado previamente.

### Sintaxis

```python
df.apply(
    func,
    axis=0,
    raw=False,
    result_type=None,
    args=(),
    **kwargs
)
```

### Parámetros

| Parámetro | Descripción |
|---|---|
| `func` | Es la función que queremos aplicar a las filas o columnas del DataFrame. Puede ser una función integrada de Python, una función de pandas o una función personalizada. |
| `axis` | Determina si la función se aplicará a las columnas (`0`) o a las filas (`1`). |
| `raw` | Si es `False`, la función recibe cada fila o columna como un objeto `Series`. Si es `True`, recibe un array de NumPy (`ndarray`). |
| `result_type` | Determina cómo se interpretarán los resultados cuando se utiliza `axis=1`. Puede ser `'expand'`, `'reduce'` o `'broadcast'`. |
| `args` | Permite pasar argumentos posicionales adicionales a la función que queremos aplicar. |
| `**kwargs` | Permite pasar argumentos adicionales con nombre a la función que queremos aplicar. |

### Parámetro `axis`

Es uno de los parámetros más importantes de `apply()`, porque determina cómo se recorrerá el DataFrame.

Tiene dos valores principales:

| Valor | Comportamiento |
|---|---|
| `axis=0` | Aplica la función a cada columna. Es el valor predeterminado. |
| `axis=1` | Aplica la función a cada fila. |

Para comprenderlo mejor, imaginemos el siguiente DataFrame:

```python
import pandas as pd

df = pd.DataFrame({
    'nombre': ['Elliot', 'Darlene', 'Angela'],
    'edad': [19, 25, 23],
    'salario': [1500, 2000, 1800]
})
```

**Con `axis=0`:**

```python
df.apply(max, axis=0)
```

Pandas aplica `max()` a cada columna por separado.

Es decir:
- En `nombre`, obtiene el valor máximo según el orden de los textos.
- En `edad`, obtiene la edad máxima.
- En `salario`, obtiene el salario máximo.

**Con `axis=1`:**

```python
df.apply(lambda fila: fila['edad'] + fila['salario'], axis=1)
```

Pandas aplica la función a cada fila.

En este caso, suma la edad y el salario de cada registro.

La función recibe una fila y accede a sus valores mediante los nombres de las columnas.

**Importante:** cuando utilizamos `axis=1`, cada vez que se ejecuta la función recibe una fila completa. Por eso podemos acceder a varias columnas de un mismo registro.

### ¿Qué es `lambda`?

Es una forma de crear funciones pequeñas y anónimas, es decir, funciones que podemos definir directamente sin utilizar `def` ni asignarles un nombre.

Su sintaxis es:

```python
lambda argumentos: expresion
```

Por ejemplo:

```python
lambda x: x * 2
```

Esta función recibe un valor `x` y devuelve su resultado multiplicado por dos.

Podemos utilizarla directamente dentro de `apply()`:

```python
df['edad'].apply(lambda x: x * 2)
```

En este caso, estamos aplicando una operación a cada valor de la columna `edad`.

**Nota:** aunque este ejemplo utiliza una columna del DataFrame, `apply()` también existe para las columnas individuales (`Series`). En esta explicación nos centramos en su uso con DataFrames.

### Aplicar una función personalizada

También podemos crear nuestra propia función utilizando `def` y después pasarla a `apply()`.

Por ejemplo:

```python
def calcular_categoria(fila):
    if fila['edad'] >= 25:
        return 'Adulto'
    else:
        return 'Joven'

df['categoria'] = df.apply(calcular_categoria, axis=1)
```

En este caso:

1. Creamos la función `calcular_categoria()`.
2. La función recibe una fila completa del DataFrame.
3. Accede al valor de la columna `edad`.
4. Evalúa la condición y devuelve una categoría.
5. `apply()` repite el proceso para cada fila.
6. El resultado se asigna a una nueva columna llamada `categoria`.

### Parámetro `raw`

Determina el tipo de objeto que recibirá la función durante su ejecución.

| Valor | Comportamiento |
|---|---|
| `False` | Es el valor predeterminado. La función recibe cada fila o columna como una `Series` de pandas. |
| `True` | La función recibe los valores como un array de NumPy (`ndarray`). |

Ejemplo:

```python
df.apply(max, axis=0, raw=False)
```

Con `raw=False`, la función recibe una `Series` por cada columna.

```python
df.apply(max, axis=0, raw=True)
```

Con `raw=True`, la función recibe un array de NumPy por cada columna.

**Importante:** `raw=True` puede ofrecer ventajas de rendimiento en determinadas operaciones numéricas, pero no permite acceder a los valores mediante nombres de columnas, porque la función recibe un array y no una `Series`.

### Parámetro `result_type`

Se utiliza cuando aplicamos una función a las filas (`axis=1`) y permite controlar cómo pandas organiza los resultados devueltos.

Tiene tres opciones principales:

| Valor | Comportamiento |
|---|---|
| `'expand'` | Expande los resultados en varias columnas cuando la función devuelve listas o estructuras similares. |
| `'reduce'` | Intenta devolver un resultado reducido, en lugar de expandirlo en varias columnas. |
| `'broadcast'` | Devuelve los resultados con la misma forma que el DataFrame original, difundiendo el resultado de cada fila a sus columnas. |

Ejemplo con `expand`:

```python
df.apply(
    lambda fila: [fila['edad'], fila['salario']],
    axis=1,
    result_type='expand'
)
```

La función devuelve una lista con dos valores por cada fila. Con `result_type='expand'`, pandas convierte esos valores en columnas separadas.

### ¿Qué ocurre si `apply()` devuelve un único valor o varios valores?

El resultado depende de lo que devuelva la función y de cómo se configure `result_type`.

- Si la función devuelve un único valor por fila, normalmente obtendremos una `Series`.
- Si devuelve varios valores, pandas puede expandirlos en un DataFrame, dependiendo de la estructura devuelta y de `result_type`.

### Diferencia entre `apply()` y `map()`

Aunque ambos permiten aplicar funciones a los datos, tienen propósitos y ámbitos diferentes:

| Método | Función |
|---|---|
| `DataFrame.apply()` | Aplica una función a cada fila o columna de un DataFrame. |
| `Series.map()` | Aplica una función o correspondencia a cada valor de una columna individual. |

### Consideración sobre el rendimiento

Aunque `apply()` es muy flexible, no siempre es la opción más eficiente.

Cuando trabajamos con operaciones numéricas, pandas suele ofrecer operaciones vectorizadas que pueden ejecutarse más rápido que recorrer las filas utilizando `apply()`.

Por ejemplo:

```python
# Utilizando apply()
df['salario_doble'] = df['salario'].apply(lambda x: x * 2)

# Utilizando una operación vectorizada
df['salario_doble'] = df['salario'] * 2
```

Ambas instrucciones producen el mismo resultado, pero la segunda suele ser más eficiente.

**En resumen:** `apply()` permite ejecutar funciones sobre las filas o columnas de un DataFrame. Sus parámetros más importantes son `func` y `axis`; los demás permiten controlar cómo se reciben los datos y cómo se organizan los resultados.


````python
# apply() aplicar función a la columna que desees 
df['clasificacion'] = df['magnitud'].apply(clasificar_magnitud)
````
