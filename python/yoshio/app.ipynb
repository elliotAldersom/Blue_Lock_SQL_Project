```python
df['id'] = df['id'].astype(int) # cambiar el tipo de dato de column id a entero
```

```python
df['latitud'] = pd.to_numeric(df['latitud'], errors='coerce') # 'Hola' -> NaN
df['longitud'] = pd.to_numeric(df['longitud'], errors='coerce') # 'Hola' -> NaN
df['profundidad'] = pd.to_numeric(df['profundidad'], errors='coerce') # 'Hola' -> NaN
df['magnitud'] = pd.to_numeric(df['magnitud'], errors='coerce') # 'Hola' -> NaN
```

```python
df.isnull().sum()  # no se encontró nulos al cambiar al tipo de dato de cada columna
```

### Modificar formato de `fecha_utc`

```python
df['fecha_utc'].head()
```

> format=%Y%m%d

```python
df['fecha_utc'] = pd.to_datetime(df['fecha_utc'].astype(str), format='%Y%m%d', errors='coerce')
```

```python
df['fecha_utc'].head()
```

### Formato a `hoa_utc
> format = '%H%M%S`

```python
df['hora_utc'].head()
```

```python
# pasar a str y usar zfill para relleno (similar a padStart de JS )
df['hora_utc'] = df['hora_utc'].astype(str).str.zfill(6)
```

```python
# para que salga la hora sin fecha => dt.time 
df['hora_utc'] = pd.to_datetime(df['hora_utc'].astype(str), format='%H%M%S', errors='coerce').dt.time
```

```python
df['hora_utc'].head()
```

```python
# Concatenamos ambos formatos y creamos una nueva columna.
df['fecha_hora_utc'] = pd.to_datetime(df['fecha_utc'].astype(str) + ' ' + df['hora_utc'].astype(str), errors='coerce')
```

```python
# separar años, meses, dias y ponerlos en una columna a cada uno 
df['year'] = df['fecha_utc'].dt.year 
df['month'] = df['fecha_utc'].dt.month 
df['day'] = df['fecha_utc'].dt.day 
```

```python
df['monht_name'] = df['fecha_utc'].dt.month_name()
df['day_name'] = df['fecha_utc'].dt.day_name()
```

```python
# funcion para determinar la categoría de la magnitud del sismo
def clasificar_magnitud(magnitud): 
    if(magnitud  < 3.9): return 'Menor'
    elif(magnitud < 4.9): return 'Ligero'
    elif(magnitud < 5.9): return 'Moredaro'
    elif(magnitud <6.9): return 'Fuerte'
    elif(magnitud <7.9): return 'Mayor'
    elif(magnitud <9.9): return 'Catastrófico'
```

```python
# apply() aplicar función a la columna que desees 
df['clasificacion'] = df['magnitud'].apply(clasificar_magnitud)
```
